# Perkalian Silang (Cross Product)



Cross product adalah operasi antara dua vektor dalam ruang tiga dimensi yang menghasilkan vektor baru yang tegak lurus terhadap kedua vektor awal.

Sifat Sifat perkalian silang :


Rumus :

$$
\vec{A} \times \vec{B} =
\begin{vmatrix}
\hat{i} & \hat{j} & \hat{k} \\
A_x & A_y & A_z \\
B_x & B_y & B_z \\
\end{vmatrix}
=
(A_y B_z - A_z B_y)\hat{i}
- (A_x B_z - A_z B_x)\hat{j}
+ (A_x B_y - A_y B_x)\hat{k}
$$


### **Contoh Soal**
$$
\vec{A} = \begin{bmatrix} 2 \\ -1 \\ 3 \end{bmatrix}, \quad
\vec{B} = \begin{bmatrix} 1 \\ 4 \\ -2 \end{bmatrix}
$$

###Tentukan hasil dari
$$
 ( \vec{A} \times \vec{B} ):
$$


$$
\vec{A} \times \vec{B} =
\begin{vmatrix}
\hat{i} & \hat{j} & \hat{k} \\
2 & -1 & 3 \\
1 & 4 & -2 \\
\end{vmatrix}
$$


$$
= \hat{i}((-1)(-2) - (3)(4))
- \hat{j}((2)(-2) - (3)(1))
+ \hat{k}((2)(4) - (-1)(1))
$$

$$
= \hat{i}(2 - 12) - \hat{j}(-4 - 3) + \hat{k}(8 + 1)
$$

$$
= -10\hat{i} + 7\hat{j} + 9\hat{k}
$$

$$
\Rightarrow \vec{A} \times \vec{B} = \begin{bmatrix} -10 \\ 7 \\ 9 \end{bmatrix}
$$


buat contoh soal c(u x v) = (cu) x v  = u x (cu) c adalah konstanta

$$
\vec{u} = \begin{bmatrix} 2 \\ -1 \\ 3 \end{bmatrix}, \quad
\vec{v} = \begin{bmatrix} 1 \\ 4 \\ -2 \end{bmatrix}, \quad
c = 3
$$

Tentukan hasil dari

$$
\vec{u} \times \vec{v} =
\begin{vmatrix}
\hat{i} & \hat{j} & \hat{k} \\
2 & -1 & 3 \\
1 & 4 & -2 \\
\end{vmatrix}
$$

$$
= \hat{i}((-1)(-2) - (3)(4))
- \hat{j}((2)(-2) - (3)(1))
+ \hat{k}((2)(4) - (-1)(1))
$$

$$
= \hat{i}(2 - 12) - \hat{j}(-4 - 3) + \hat{k}(8 + 1)
$$

$$
= -10\hat{i} + 7\hat{j} + 9\hat{k}
$$

$$
\Rightarrow \vec{u} \times \vec{v} = \begin{bmatrix} -10 \\ 7 \\ 9 \end{bmatrix}
$$

Sekarang, hitung

$$
c(\vec{u} \times \vec{v}) = 3 \begin{bmatrix} -10 \\ 7 \\ 9 \end{bmatrix} = \begin{bmatrix} -30 \\ 21 \\ 27 \end{bmatrix}
$$


Hitung :

$$
c\vec{u} = 3 \begin{bmatrix} 2 \\ -1 \\ 3 \end{bmatrix} = \begin{bmatrix} 6 \\ -3 \\ 9 \end{bmatrix}
$$

$$
(c\vec{u}) \times \vec{v} =
\begin{vmatrix}
\hat{i} & \hat{j} & \hat{k} \\
6 & -3 & 9 \\
1 & 4 & -2 \\
\end{vmatrix}
= \hat{i}((-3)(-2) - (9)(4)) - \hat{j}((6)(-2) - (9)(1)) + \hat{k}((6)(4) - (-3)(1))
$$

$$
= \hat{i}(6 - 36) - \hat{j}(-12 - 9) + \hat{k}(24 + 3)
$$

$$
= -30\hat{i} + 21\hat{j} + 27\hat{k}
$$

$$
= \begin{bmatrix} -30 \\ 21 \\ 27 \end{bmatrix}
$$

Hitung :

$$
c\vec{v} = 3 \begin{bmatrix} 1 \\ 4 \\ -2 \end{bmatrix} = \begin{bmatrix} 3 \\ 12 \\ -6 \end{bmatrix}
$$

$$
\vec{u} \times (c\vec{v}) =
\begin{vmatrix}
\hat{i} & \hat{j} & \hat{k} \\
2 & -1 & 3 \\
3 & 12 & -6 \\
\end{vmatrix}
= \hat{i}((-1)(-6) - (3)(12)) - \hat{j}((2)(-6) - (3)(3)) + \hat{k}((2)(12) - (-1)(3))
$$

$$
= \hat{i}(6 - 36) - \hat{j}(-12 - 9) + \hat{k}(24 + 3)
$$

$$
= -30\hat{i} + 21\hat{j} + 27\hat{k}
$$

$$
= \begin{bmatrix} -30 \\ 21 \\ 27 \end{bmatrix}
$$

$$
\boxed{
c(\vec{u} \times \vec{v}) = (c\vec{u}) \times \vec{v} = \vec{u} \times (c\vec{v}) = \begin{bmatrix} -30 \\ 21 \\ 27 \end{bmatrix}
}
$$
