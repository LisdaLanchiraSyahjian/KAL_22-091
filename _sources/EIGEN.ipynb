# Eigen value dan Eigen Vektor


## Definisi Eigen value dan Eigen Vektor
- Eigen value adalah Bilangan skalar
𝜆 yang memenuhi:

$$
A \mathbf{v} = \lambda \mathbf{v}
$$

- Eigen vektor adalah Vektor tak nol
𝑣
⃗
v
  yang arahnya tidak berubah saat dikalikan dengan matriks
𝐴
A, hanya panjangnya yang berubah sesuai
𝜆.

## 📌 Soal

Diketahui matriks:

$$
A = \begin{bmatrix} 4 & 1 \\ 2 & 3 \end{bmatrix}
$$

Tentukan **eigenvalue** dan **eigenvector** dari matriks \( A \).

---

## ✏️ Langkah 1: Mencari Eigenvalue

Gunakan rumus:

$$
\det(A - \lambda I) = 0
$$

$$
\Rightarrow \det \left( \begin{bmatrix} 4 - \lambda & 1 \\ 2 & 3 - \lambda \end{bmatrix} \right) = 0
$$

$$
(4 - \lambda)(3 - \lambda) - (2 \cdot 1) = 0
$$

$$
(12 - 4\lambda - 3\lambda + \lambda^2) - 2 = 0
$$

$$
\lambda^2 - 7\lambda + 10 = 0
$$

$$
(\lambda - 5)(\lambda - 2) = 0
$$

$$
\Rightarrow \lambda_1 = 5, \quad \lambda_2 = 2
$$

---

## ✏️ Langkah 2: Mencari Eigenvector

### Untuk \( \lambda = 5 \):

$$
A - 5I = \begin{bmatrix} -1 & 1 \\ 2 & -2 \end{bmatrix}
$$

$$
\Rightarrow \begin{bmatrix} -1 & 1 \\ 2 & -2 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} = \begin{bmatrix} 0 \\ 0 \end{bmatrix}
$$

$$
\Rightarrow -x + y = 0 \Rightarrow y = x
$$

$$
\Rightarrow \vec{v}_1 = \begin{bmatrix} 1 \\ 1 \end{bmatrix}
$$

---

### Untuk \( \lambda = 2 \):

$$[
A - 2I = \begin{bmatrix} 2 & 1 \\ 2 & 1 \end{bmatrix}
$$

$$
\Rightarrow \begin{bmatrix} 2 & 1 \\ 2 & 1 \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix} = \begin{bmatrix} 0 \\ 0 \end{bmatrix}
$$

$$
\Rightarrow 2x + y = 0 \Rightarrow y = -2x
$$

$$
\Rightarrow \vec{v}_2 = \begin{bmatrix} 1 \\ -2 \end{bmatrix}
$$

---

## ✅ Jawaban Akhir

- **Eigenvalue**:
  $$
  \lambda_1 = 5, \quad \lambda_2 = 2
  $$

- **Eigenvector**:
  $$
  \vec{v}_1 = \begin{bmatrix} 1 \\ 1 \end{bmatrix}, \quad
  \vec{v}_2 = \begin{bmatrix} 1 \\ -2 \end{bmatrix}
  $$


```python
import numpy as np

# Matriks A
A = np.array([[4, 1],
              [2, 3]])

# Menghitung eigenvalue dan eigenvector
eigenvalues, eigenvectors = np.linalg.eig(A)

# Tampilkan hasil
print("Matriks A:")
print(A)

print("\nEigenvalue:")
print(eigenvalues)

print("\nEigenvector (setiap kolom adalah eigenvector):")
print(eigenvectors)
```
